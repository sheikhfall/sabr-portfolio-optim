```python
import pyarrow.parquet as pq, json, itertools

for f in ["../data/historical_prices.parquet", "../data/market_cap.parquet", "../data/model_signal.parquet"]:
    pf = pq.ParquetFile(f)
    print("\n==", f, "| rows:", pf.metadata.num_rows)
    print(pf.schema_arrow)
    print(pf.read_row_group(0).slice(0, 5).to_pandas())

d = json.load(open("../data/compliance_dict.json"))
print("\n== compliance_dict.json | type:", type(d).__name__, "| len:", len(d))
print(list(itertools.islice(d.items(), 3)) if isinstance(d, dict) else d[:3])
```
